##  1. INDENTATION — Python's Way of Structuring Code

Most other programming languages uses braces {} to determine block structure but python uses whitespace and indentation to determine it.

Indentation refers to the leading whitespace (spaces or tabs) at the beginning of a line of code.

* It tells the Python interpreter which statements belong to a specific loop, conditional block, class, or function.
 * Skipping or mismatching indentation will trigger an IndentationError or TabError.

```python
int n, r;
n = 9;
r = 1;
while (n > 0) {
  r *= n;
  n--;
}
```

```python
n = 9
r = 1
while n > 0:
  r *= n # r = r * n
  n -= 1 # n - 1
```

## 6. PRACTICAL EXAMPLE — Putting It All Together


```python
def calculate_bmi(weight, height):
  """
    Calculate Body Mass Index (BMI).

    Parameters:
        weight (float): Weight in kilograms
        height (float): Height in meters

    Returns:
        tuple: (bmi_value, category)
  """
  # Calculate BMI using the standrad formula
  bmi = weight / (height ** 2)

  # Determine the BMI category
  if bmi < 18.5:
    category = "Underweight"
  elif bmi < 25:
    category = "Normal"
  elif bmi < 30:
    category = "Overweight"
  else:
    category = "Obese"

  return bmi, category
```

```python
weight_kg = 70
height_m = 1.67

bmi, cat = calculate_bmi(weight_kg, height_m)
bmi
cat
```
